```python
dbutils.widgets.text('filename', '')
```

```python
p_filename = dbutils.widgets.get('filename')
```

### **DATA READING**

```python
df = spark.readStream.format('cloudFiles')\
    .option('cloudFiles.format', 'parquet')\
    .option('cloudFiles.schemaLocation', f'abfss://bronze@databricksretailproject.dfs.core.windows.net/checkpoint_{p_filename}')\
    .load(f'abfss://source@databricksretailproject.dfs.core.windows.net/{p_filename}')
```

```python
df.writeStream.format('parquet')\
    .outputMode('append')\
    .option('checkpointLocation', f'abfss://bronze@databricksretailproject.dfs.core.windows.net/checkpoint_{p_filename}')\
    .option('path', f'abfss://bronze@databricksretailproject.dfs.core.windows.net/{p_filename}')\
    .trigger(once=True)\
    .start()
```
